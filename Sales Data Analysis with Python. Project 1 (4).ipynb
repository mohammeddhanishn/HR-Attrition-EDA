```python
Q1 = df['MonthlyIncome'].quantile(0.25)

Q3 = df['MonthlyIncome'].quantile(0.75)

IQR = Q3 - Q1

Lower = Q1 - 1.5 * IQR

Upper = Q3 + 1.5 * IQR

print("Lower Limit:", Lower)
print("Upper Limit:", Upper)
```

```python
outliers = df[
    (df['MonthlyIncome'] < Lower) |
    (df['MonthlyIncome'] > Upper)
]

print("Number of Outliers:", len(outliers))
```

```python
corr = df.corr(numeric_only=True)

upper = corr.where(np.triu(np.ones(corr.shape), k=1).astype(bool))

top10 = (
    upper.stack()
         .rename("correlation")
         .to_frame()
         .assign(strength=lambda x: x["correlation"].abs())
         .sort_values("strength", ascending=False)
         .head(10)
         .reset_index()
)

top10["pair"] = top10["level_0"] + " ↔ " + top10["level_1"]

plt.figure(figsize=(10, 6))
sns.barplot(data=top10, x="correlation", y="pair", hue="correlation",
            palette="coolwarm", legend=False)
plt.title("Top 10 Strongest Correlations")
plt.xlabel("Correlation")
plt.ylabel("Feature pair")
plt.tight_layout()
plt.show()
```
